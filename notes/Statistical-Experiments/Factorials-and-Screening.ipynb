# Factorials & Screening

Study factors together, interpret interactions, and use fractions and follow-up runs to learn efficiently.

## Topics

::::{grid} 1 1 2 2

:::{grid-item-card} [Factorial Experiments & Interaction](Factorial-Experiments.ipynb)

An interaction means the effect of one factor depends on another. A main effect can hide that dependence.
:::

:::{grid-item-card} [Two-Level Factorial Designs](Two-Level-Factorials.ipynb)

Coded sign columns turn factorial effects into orthogonal contrasts. Distinguish an effect from its regression coefficient.
:::

:::{grid-item-card} [Fractional Factorial Designs](Fractional-Factorials.ipynb)

Fewer runs mean aliased effects. Interpret an estimated contrast together with its complete alias set.
:::

:::{grid-item-card} [Saturated Designs & Foldovers](Saturated-Designs-and-Foldovers.ipynb)

A saturated fit has no residual degrees of freedom. Follow-up runs should resolve specific ambiguities in the alias structure.
:::

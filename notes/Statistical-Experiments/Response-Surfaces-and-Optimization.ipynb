# Response Surfaces & Optimization

Move from detecting curvature to locating useful operating conditions and reducing sensitivity to noise.

## Topics

::::{grid} 1 1 2 2

:::{grid-item-card} [Center Points & Curvature](Center-Points-and-Curvature.ipynb)

Center points can detect aggregate curvature and estimate pure error. They cannot separately identify every quadratic term.
:::

:::{grid-item-card} [Response Surfaces & Steepest Ascent](Response-Surfaces-and-Steepest-Ascent.ipynb)

Fit locally, move in a promising direction, observe the response, and recenter. A local plane is not a global optimization model.
:::

:::{grid-item-card} [Quadratic Response Surfaces](Quadratic-Response-Surfaces.ipynb)

Solve for a stationary point, then classify it using curvature. A stationary point is not automatically a useful maximum.
:::

:::{grid-item-card} [Robust Parameter Design](Robust-Parameter-Design.ipynb)

Choose settings that achieve the desired mean while controlling sensitivity to noise. The best mean and the smallest variance may occur at different settings.
:::

# Fractional Factorial Designs

A **regular $2^{k-p}$ fractional factorial** studies $k$ two-level factors in $2^{k-p}$ runs, a $1/2^p$ fraction of the full factorial. It saves runs by accepting known alias relationships among effects.

A fraction is a carefully constructed design, not an arbitrary subset of convenient settings. It is most useful when a limited set of effects is scientifically plausible and further experimentation is possible.

## Why Fractions Can Be Useful

| Idea | Meaning | Limitation |
|---|---|---|
| Effect sparsity | Relatively few effects may dominate the response. | A working assumption, not a guarantee. |
| Effect hierarchy | Low-order effects are often prioritized. | A high-order interaction is not automatically zero. |
| Projection | A design can be balanced in a smaller set of factors. | Ignored factors must be irrelevant for the intended reduced analysis. |
| Sequential experimentation | Add runs to resolve ambiguities. | Later runs must be combined with attention to process changes. |

An aliased contrast can be large because of any component in its alias set. Calling it a pure main effect requires assumptions about the other components.

## Constructing a Half-Fraction

To study four factors in eight runs, start with a full $2^3$ design in A, B, C and set

$$D=ABC.$$

This is a **design generator**: the D setting in each row is the product of the A, B, C signs. Multiplying by D gives the **defining relation**

$$I=ABCD.$$

| A | B | C | D=ABC | Treatment label |
|---:|---:|---:|---:|---|
| -1 | -1 | -1 | -1 | $(1)$ |
| +1 | -1 | -1 | +1 | $ad$ |
| -1 | +1 | -1 | +1 | $bd$ |
| +1 | +1 | -1 | -1 | $ab$ |
| -1 | -1 | +1 | +1 | $cd$ |
| +1 | -1 | +1 | -1 | $ac$ |
| -1 | +1 | +1 | -1 | $bc$ |
| +1 | +1 | +1 | +1 | $abcd$ |

All these rows have ABCD sign $+1$. The complementary fraction uses $D=-ABC$, giving ABCD sign $-1$. Both have the same structural quality, but the signs in their alias relations differ.

The sign table specifies which runs to perform. Randomize their execution order subject to actual constraints.

## Alias Sets

Multiply the defining relation by the desired term and cancel squared letters:

$$A=A(ABCD)=BCD,\qquad AB=AB(ABCD)=CD.$$

The complete distinct alias pairs are

| Column | Same column on the principal fraction |
|---|---|
| I | ABCD |
| A | BCD |
| B | ACD |
| C | ABD |
| D | ABC |
| AB | CD |
| AC | BD |
| AD | BC |

For example, if the full response model contains coefficients $\beta_A$ and $\beta_{BCD}$, the estimated A-column coefficient has expectation

$$E(\widehat b_A)=\beta_A+\beta_{BCD}.$$

On the complementary fraction, it instead estimates $\beta_A-\beta_{BCD}$. Here $b_A$ denotes a fitted coefficient for the observed column; $\beta_A$ denotes a component of the full underlying model. Under the usual $\pm1$ convention, factorial effects are twice these coefficients.

:::{dropdown} Numerical Alias Example

Suppose the mean response is

$$\mu=20+2A+3BCD.$$

On $D=ABC$, the BCD column equals A, so the observed mean is $20+5A$. The fitted A-column coefficient is 5, although the true A coefficient is only 2.

On the complementary fraction, $BCD=-A$, and the fitted coefficient is $2-3=-1$.

Together,

$$\frac{5+(-1)}2=2,\qquad\frac{5-(-1)}2=3.$$

The two fractions supply the information needed to distinguish these terms. This illustration is noiseless; with data, both recovered estimates have sampling uncertainty.
:::

## Resolution

The **resolution** of a regular fraction is the length of the shortest nonidentity word in its complete defining relation. A word is a product of distinct factor letters after cancellations.

| Resolution | Low-order aliasing pattern |
|---|---|
| III | Main effects are not aliased with other main effects, but can be aliased with two-way interactions. |
| IV | Main effects are clear of two-way interactions; two-way interactions can be aliased with one another. |
| V | Main effects and two-way interactions are not aliased with one another; two-way terms can be aliased with three-way terms. |

“Clear” here refers to the specified low-order effects, not absence of all higher-order aliases.

The design $I=ABCD$ has resolution IV. The design $I=ABCDE$ has resolution V. A half-fraction has resolution $k$ **when its defining word includes all $k$ factors**; an arbitrary half-fraction need not have this maximum resolution.

## Quarter-Fractions & the Complete Defining Group

For a $2^{5-2}$ design, use A, B, C as base factors and set

$$D=AB,\qquad E=AC.$$

The generators give $I=ABD$ and $I=ACE$. Their product supplies a third word:

$$ABD\cdot ACE=BCDE.$$

Therefore,

$$\boxed{I=ABD=ACE=BCDE}.$$

It has resolution III. Multiplying the entire relation by A gives

$$A=BD=CE=ABCDE.$$

Thus an A-column coefficient represents $\beta_A+\beta_{BD}+\beta_{CE}+\beta_{ABCDE}$ for the principal fraction.

Other useful alias sets include

$$B=AD=ABCE=CDE,$$

$$C=ABCD=AE=BDE,$$

$$BC=ACD=ABE=DE.$$

For $p$ independent generators, there are $2^p$ words including I, or $2^p-1$ nonidentity defining words. Each complete alias class contains $2^p$ effects. Complementary generator signs can introduce minus signs in these relationships.

:::{note}
Use **independent** generators and distinct, appropriate columns for the added factors. Duplicate or dependent generators do not provide the intended number of independent restrictions. Verify the full defining group before assigning scientific factor names.
:::

## Selecting Generators: Resolution & Aberration

First identify effects that must be protected. Among candidate regular designs with suitable run counts, compare resolution and the entire word-length pattern.

Let $A_j$ count defining words of length $j$. **Minimum aberration** compares $(A_1,A_2,\ldots)$ lexicographically: minimize the first count at which designs differ. At tied resolution, this starts by preferring fewer words of the shortest length, then continues to longer lengths if needed.

:::{dropdown} Two 32-Run Designs for Eight Factors

Use base factors A–E and generate F, G, H.

| Candidate | Generators | Nonidentity word lengths |
|---|---|---|
| 1 | $F=ABC$, $G=ABD$, $H=ACD$ | Seven words of length 4. |
| 2 | $F=ABC$, $G=ABD$, $H=ACDE$ | Three words of length 4; four of length 5. |

Both are $2^{8-3}$ resolution-IV designs. Candidate 2 has lower aberration because $A_4=3$ rather than 7. Fewer short defining words mean fewer low-order alias conflicts.

This is not a claim that candidate 2 is best for every scientific question. Factor labeling and prior knowledge can make a particular alias pair especially costly. Inspect the effects that matter for the actual experiment.
:::

## Projection: What Happens When Factors Are Ignored?

A regular resolution-$R$ design is balanced on every set of at most $R-1$ factors. Projecting onto such a set yields every combination equally often.

For the maximum-resolution half-fraction $I=ABCD$, projection onto any three factors is a full $2^3$ design. Onto any two factors, each of the four settings appears twice.

This is a property of the design table; it does not make the different original settings genuine replicates of the same complete treatment.

:::{warning}
Two runs with the same A and B values can still differ in C and D. Treating them as replicates in a reduced A–B analysis assumes C, D, and their relevant interactions do not affect the response. A nonsignificant screen alone does not prove that assumption.
:::

## Estimation, Error & Follow-Up

For mutually orthogonal estimable columns in a regular fraction, a column coefficient can be computed as $X_j^\mathsf Ty/N$. Its interpretation is an **alias combination**, not necessarily an isolated effect.

Residual df depends on the fitted model. Eight runs with an intercept and four main effects leave three residual df, but these remaining contrasts can contain interactions. They are not pure replication error.

Ways to improve the evidence include independently replicating settings, obtaining appropriate external error information, or adding a designed follow-up that changes the aliases. Report the alias assumptions along with effect estimates.

Combining principal and complementary halves recovers the full factorial treatment layout. If the halves were run in different phases and a phase effect is fitted, that phase indicator is confounded with the original defining-word interaction, such as ABCD. Lower-order aliases can be resolved without claiming that **every** effect is separately estimable from phase.

## Check Generators & Aliases in Python

The cell uses letter cancellation to compute defining words, alias sets, resolution, and word-length counts. It also checks the noiseless numerical example against direct regression.

```python
from collections import Counter
import numpy as np

def multiply(u, v):
    # Repeated letters cancel: symmetric difference of letter sets.
    return frozenset(u) ^ frozenset(v)

def defining_group(generators):
    words = {frozenset()}
    for word in generators:
        words |= {multiply(w, word) for w in list(words)}
    assert len(words) == 2**len(generators), "Generators must be independent"
    return words

def show(word):
    return "".join(sorted(word)) or "I"

def aliases(term, words):
    return sorted((show(multiply(term, w)) for w in words), key=lambda s:(len(s),s))

quarter = defining_group(["ABD", "ACE"])
print("Quarter-fraction defining words:", sorted(map(show, quarter)))
print("A aliases:", aliases("A", quarter))
for name, gens in [
    ("Candidate 1", ["ABCF", "ABDG", "ACDH"]),
    ("Candidate 2", ["ABCF", "ABDG", "ACDEH"]),
]:
    group = defining_group(gens)
    lengths = Counter(len(w) for w in group if w)
    print(name, "resolution", min(lengths), "word counts", dict(sorted(lengths.items())))

base = np.array([[1 if (i >> j) & 1 else -1 for j in range(3)]
                 for i in range(8)])
A, B, C = base.T
estimates = []
for sign in [1, -1]:
    D = sign*A*B*C
    y = 20 + 2*A + 3*B*C*D
    X = np.column_stack([np.ones(8), A, B, C, D])
    estimates.append(np.linalg.lstsq(X, y, rcond=None)[0][1])
print("A-column estimates in the two halves:", np.round(estimates, 6))
print("Recovered A and BCD coefficients:", np.mean(estimates), (estimates[0]-estimates[1])/2)
assert np.allclose(estimates, [5, -1])
```

## Quick Review

:::{dropdown} Questions & Answers

**Is resolution the number of runs?** No. It is the shortest nonidentity defining-word length.

**Why do two generators produce three nonidentity defining words?** Their product is also a defining word.

**Does a large estimated main-effect column prove that main effect is active?** No. Another term in its alias set can produce the same signal.

**Are projected repeat settings automatically pure replicates?** No. They can differ in omitted factors.

**Can a fraction support a unique full factorial model?** No. It provides fewer independent columns than that model requires.
:::

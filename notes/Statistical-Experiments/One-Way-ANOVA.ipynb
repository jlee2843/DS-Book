# One-Way ANOVA

Compare several treatment means by separating between-treatment variation from within-treatment error. This page covers the global test, equivalent model parameterizations, diagnostics, and the alternative framework for binary outcomes. Numerical examples use illustrative practice data.

## One-Way ANOVA

ANOVA compares one factor with $m$ treatment levels. The standard normal-response model is

$$Y_{ij}=\mu_i+\varepsilon_{ij},\qquad
\varepsilon_{ij}\overset{\mathrm{iid}}{\sim}N(0,\sigma^2),\qquad
N=\sum_{i=1}^m n_i.$$

$$H_0:\mu_1=\cdots=\mu_m,\qquad
H_A:\text{not all treatment means are equal}.$$

The alternative does not say that every pair differs.

### Decomposing Variation

Define the group means $\bar Y_{i\cdot}$ and the **observation-weighted** grand mean

$$\bar Y_{\cdot\cdot}=\frac1N\sum_i n_i\bar Y_{i\cdot}.$$

$$\begin{aligned}
SS_T&=\sum_i\sum_j(Y_{ij}-\bar Y_{\cdot\cdot})^2,\\
SS_{\mathrm{Tr}}&=\sum_i n_i(\bar Y_{i\cdot}-\bar Y_{\cdot\cdot})^2,\\
SS_E&=\sum_i\sum_j(Y_{ij}-\bar Y_{i\cdot})^2.
\end{aligned}$$

$$\boxed{SS_T=SS_{\mathrm{Tr}}+SS_E}$$

**Treatment variation** measures separation among group means. **Error variation** measures spread within each group. Some notation calls $SS_{\mathrm{Tr}}$ “condition SS” or $SS_C$.

:::{dropdown} Why the Sums of Squares Add Up

Write each deviation as

$$Y_{ij}-\bar Y_{\cdot\cdot}
=(Y_{ij}-\bar Y_{i\cdot})+(\bar Y_{i\cdot}-\bar Y_{\cdot\cdot}).$$

Square and sum. The cross term vanishes because, in each group,

$$\sum_j(Y_{ij}-\bar Y_{i\cdot})=0.$$

The remaining terms are exactly $SS_E$ and $SS_{\mathrm{Tr}}$. The decomposition holds even when group sizes differ.
:::

### ANOVA Table

| Source | Sum of squares | df | Mean square | Test |
|---|---|---|---|---|
| Treatment | $SS_{\mathrm{Tr}}$ | $m-1$ | $MS_{\mathrm{Tr}}=SS_{\mathrm{Tr}}/(m-1)$ | $F=MS_{\mathrm{Tr}}/MSE$ |
| Error | $SS_E$ | $N-m$ | $MSE=SS_E/(N-m)$ | |
| Total | $SS_T$ | $N-1$ | | |

Under $H_0$ and the model assumptions,

$$F\sim F_{m-1,N-m},\qquad p=P(F_{m-1,N-m}\geq F_{\mathrm{obs}}).$$

Both mean squares estimate $\sigma^2$ under the null. Separated treatment means increase the numerator, so rejection is in the **right tail**. This is a test about means, unlike the two-sample variance-ratio test.

For $m=2$, this ANOVA test equals the **two-sided pooled $t$ test**, with $F=t^2$. It is not generally equivalent to Welch's test.

:::{dropdown} Worked ANOVA Example

| Treatment | Observations | Mean |
|---|---|---:|
| A | 5, 7, 6, 6 | 6 |
| B | 8, 9, 7, 8 | 8 |
| C | 10, 12, 11, 11 | 11 |

Here $N=12$, $m=3$, and $\bar y_{\cdot\cdot}=25/3=8.3333$.

$$SS_{\mathrm{Tr}}=4[(6-25/3)^2+(8-25/3)^2+(11-25/3)^2]=50.6667.$$

Each group contributes $2$ to within-group SS, so $SS_E=6$.

| Source | SS | df | MS | $F$ |
|---|---:|---:|---:|---:|
| Treatment | 50.6667 | 2 | 25.3333 | 38 |
| Error | 6 | 9 | 0.6667 | |
| Total | 56.6667 | 11 | | |

$$p=P(F_{2,9}\geq38)\approx0.00004090.$$

There is evidence that the population means are not all equal under the model. The global test does not identify which comparisons drive the result.
:::

## Equivalent Model Parameterizations

The following models have the same fitted group means, residuals, and overall ANOVA test.

| Model | Form | Interpretation |
|---|---|---|
| Means | $Y_{ij}=\mu_i+\varepsilon_{ij}$ | A parameter for each group mean. |
| Effects | $Y_{ij}=\mu+\tau_i+\varepsilon_{ij}$, $\sum_i\tau_i=0$ | Mean of treatment means plus deviations. |
| Reference-group regression | $Y=\beta_0+\sum_{k=2}^m\beta_{k-1}I(\text{group}=k)+\varepsilon$ | Reference mean plus differences from it. |

In the effects model with an **unweighted sum-to-zero constraint**,

$$\mu=\frac1m\sum_i\mu_i,\qquad
\widehat\mu=\frac1m\sum_i\bar Y_{i\cdot},\qquad
\widehat\tau_i=\bar Y_{i\cdot}-\widehat\mu.$$

For unequal group sizes, this average of group means is not necessarily the observation-weighted grand mean used in the ANOVA decomposition. With balanced groups, they coincide.

For A, B, C with A as the reference,

$$Y=\beta_0+\beta_1I(B)+\beta_2I(C)+\varepsilon.$$

The worked example gives $\widehat\beta_0=6$, $\widehat\beta_1=2$, and $\widehat\beta_2=5$. The global null is $\beta_1=\beta_2=0$, with no requirement that the intercept be zero.

**Avoid the dummy-variable trap:** with an intercept, include only $m-1$ group indicators. Alternatively, use all $m$ indicators without an intercept.

## Residual Diagnostics & Other Response Types

For one-way ANOVA, fitted values are group means and residuals are

$$\widehat Y_{ij}=\bar Y_{i\cdot},\qquad e_{ij}=Y_{ij}-\bar Y_{i\cdot}.$$

| Check | What to look for | Implication |
|---|---|---|
| Assignment and measurement design | Repeated units, clusters, or matched observations. | Independence must come from the design, not a normality plot. |
| Residual Q–Q plot | Strong curvature or extreme tails. | Normal-theory inference may be unreliable, especially with small samples. |
| Residuals vs. fitted values | Changing spread across fitted groups. | Common variance may be unsuitable. |
| Residuals vs. collection order | Drift, cycles, long runs. | Time structure or omitted nuisance effects. |
| Unusual observations | Recording mistakes or exceptional conditions. | Investigate context; do not remove points just to improve significance. |

Normality tests can supplement plots but do not certify a model. With unequal variances, consider a heteroscedastic analysis such as Welch's ANOVA. For strongly skewed outcomes, a meaningful transformation or a suitable response model may be preferable. Count outcomes may need a count model; binary outcomes need a proportions or binary-response analysis.

### Binary Outcomes: Comparing Several Proportions

For treatment $i$, let $O_{1i}$ be failures and $O_{2i}$ successes, with $n_i=O_{1i}+O_{2i}$. Test

$$H_0:\pi_1=\cdots=\pi_m.$$

Estimate the common success probability under the null:

$$\widehat\pi=\frac{\sum_iO_{2i}}N,\qquad
E_{1i}=n_i(1-\widehat\pi),\quad E_{2i}=n_i\widehat\pi.$$

$$X^2=\sum_{i=1}^m\sum_{r=1}^2\frac{(O_{ri}-E_{ri})^2}{E_{ri}}
\ \overset{\mathrm{approx}}{\sim}\ \chi^2_{m-1}.$$

Use the right tail. This approximation requires independent observations and sufficiently large expected counts; sparse tables call for an appropriate exact or simulation-based method. The global test does not identify which proportions differ.

## Reproduce the ANOVA Example

```python
import numpy as np
from scipy import stats
# Balanced one-way ANOVA, four observations per treatment.
y = np.array([[5, 7, 6, 6], [8, 9, 7, 8], [10, 12, 11, 11]], dtype=float)
m, n = y.shape
means, grand = y.mean(axis=1), y.mean()
ss_tr = n * np.sum((means - grand)**2)
ss_e = np.sum((y - means[:, None])**2)
df_e = y.size - m
mse = ss_e / df_e
F = (ss_tr / (m - 1)) / mse
assert np.isclose(ss_tr + ss_e, np.sum((y - grand)**2))
assert np.isclose(F, stats.f_oneway(*y).statistic)
print(f"ANOVA: F={F:.4f}, p={stats.f.sf(F, m-1, df_e):.8f}")

```

## Quick Review

:::{dropdown} Self-Check
**Does a significant global test show that every pair differs?** No. It says the population means are not all equal.

**Which variation estimates common error variance?** Within-group variation, summarized by $MSE=SS_E/(N-m)$.

**Does changing from means coding to reference coding change fitted values?** No. Equivalent parameterizations describe the same group-mean model, but their coefficients have different meanings.

**Should binary observations automatically use normal-response ANOVA?** No. Use a proportions or binary-response analysis with suitable assumptions.
:::

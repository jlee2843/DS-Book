# Blocking Designs

Blocking accounts for known nuisance variation while preserving fair treatment comparisons. This page covers complete blocks, balanced incomplete blocks, and Latin squares. Numerical examples use illustrative practice data.

## Randomized Complete Block Designs

Use a **randomized complete block design (RCBD)** when a known nuisance variable creates meaningful groups of similar units, and every treatment fits within every block.

Examples: compare production settings within each batch, or compare several conditions within each participant. **Randomize within blocks**, not across the whole experiment after forming them.

For $m$ treatments and $b$ blocks, with one observation per treatment–block cell,

$$Y_{ij}=\mu+\tau_i+\beta_j+\varepsilon_{ij},\qquad N=mb,$$

$$\sum_i\tau_i=0,\quad\sum_j\beta_j=0,\qquad
\varepsilon_{ij}\overset{\mathrm{iid}}{\sim}N(0,\sigma^2).$$

| Parameter | Meaning | Estimate in this balanced design |
|---|---|---|
| $\mu$ | Overall mean | $\bar Y_{\cdot\cdot}$ |
| $\tau_i$ | Treatment deviation | $\bar Y_{i\cdot}-\bar Y_{\cdot\cdot}$ |
| $\beta_j$ | Block deviation | $\bar Y_{\cdot j}-\bar Y_{\cdot\cdot}$ |

$$\widehat Y_{ij}=\bar Y_{i\cdot}+\bar Y_{\cdot j}-\bar Y_{\cdot\cdot},\qquad
e_{ij}=Y_{ij}-\widehat Y_{ij}.$$

This is an **additive** model. With only one observation per cell, treatment-by-block interaction cannot be separated from random error. The simple analysis assumes interaction is negligible. Repeated measurements can also require a covariance model beyond these independent errors.

### Block ANOVA

$$\begin{aligned}
SS_{\mathrm{Tr}}&=b\sum_i(\bar Y_{i\cdot}-\bar Y_{\cdot\cdot})^2,\\
SS_B&=m\sum_j(\bar Y_{\cdot j}-\bar Y_{\cdot\cdot})^2,\\
SS_E&=\sum_{i,j}(Y_{ij}-\bar Y_{i\cdot}-\bar Y_{\cdot j}+\bar Y_{\cdot\cdot})^2.
\end{aligned}$$

$$SS_T=SS_{\mathrm{Tr}}+SS_B+SS_E.$$

| Source | df | MS | Test |
|---|---|---|---|
| Treatment | $m-1$ | $SS_{\mathrm{Tr}}/(m-1)$ | $MS_{\mathrm{Tr}}/MSE$ |
| Block | $b-1$ | $SS_B/(b-1)$ | $MS_B/MSE$ |
| Error | $(m-1)(b-1)$ | $SS_E/[(m-1)(b-1)]$ | |
| Total | $mb-1$ | | |

For $H_0:\tau_1=\cdots=\tau_m=0$,

$$F_{\mathrm{Tr}}\sim F_{m-1,(m-1)(b-1)}.$$

A fixed-block test uses $F_{b-1,(m-1)(b-1)}$ under $H_0:\beta_1=\cdots=\beta_b=0$. The scientific focus is usually the treatment comparison. Blocking should be planned from knowledge of the experiment, not selected solely through a block $p$-value.

A treatment contrast uses

$$\operatorname{SE}(\widehat C)=\sqrt{\frac{MSE}{b}\sum_i c_i^2},$$

with the **block-model residual df**. For two treatments, the treatment $F$ test equals the squared paired $t$ statistic.

:::{dropdown} Worked Blocking Example: Separating Treatment from Batch

| Treatment | Batch 1 | Batch 2 | Batch 3 | Batch 4 | Mean |
|---|---:|---:|---:|---:|---:|
| A | 8 | 12 | 9 | 14 | 10.75 |
| B | 10 | 13 | 13 | 16 | 13.00 |
| C | 13 | 16 | 14 | 18 | 15.25 |

The grand mean is $13$. For A in Batch 1,

$$\widehat y_{A1}=10.75+10.3333-13=8.0833,\qquad e_{A1}=-0.0833.$$

| Source | SS | df | MS | $F$ |
|---|---:|---:|---:|---:|
| Treatment | 40.5000 | 2 | 20.2500 | 42.8824 |
| Batch | 52.6667 | 3 | 17.5556 | 37.1765 |
| Error | 2.8333 | 6 | 0.4722 | |
| Total | 96.0000 | 11 | | |

The treatment-test $p$-value is approximately $0.000280$.

Ignoring batches puts their variation into the one-way residual:

$$MSE_{\mathrm{unblocked}}=\frac{52.6667+2.8333}{9}=6.1667.$$

Blocking reduces the error estimate to $0.4722$ here. It also spends three residual degrees of freedom estimating batch effects. A block variable with little explanatory value need not improve precision.
:::

The mean-difference estimates and simple orthogonal SS formulas above rely on the **complete balanced layout**. Missing cells or unequal replication require fitting the appropriate linear model; marginal mean subtraction is not generally enough.

## Balanced Incomplete Block Designs

When each block can hold only $k<v$ treatments, a **balanced incomplete block design (BIBD)** spreads comparisons evenly.

| Symbol | Meaning |
|---|---|
| $v$ | Number of treatments. |
| $b$ | Number of blocks. |
| $k$ | Treatments per block, each occurring once in that block. |
| $r$ | Blocks containing each treatment. |
| $\lambda$ | Blocks containing each pair of distinct treatments. |

Counting all assignments and treatment co-occurrences gives

$$bk=vr,\qquad r(k-1)=\lambda(v-1).$$

These are necessary conditions, not a guarantee that a design exists for arbitrary parameters. Under the usual additive model, balance makes treatment pair comparisons equally precise.

:::{dropdown} Small BIBD: Four Treatments, Two per Block

Use blocks AB, AC, AD, BC, BD, and CD.

- $v=4$, $b=6$, $k=2$.
- Each treatment appears $r=3$ times.
- Every pair appears together $\lambda=1$ time.

$$bk=6(2)=4(3)=vr,$$

$$r(k-1)=3(1)=1(3)=\lambda(v-1).$$

There are 12 observations. For this connected additive fixed-effects design, residual df is $12-4-6+1=3$.

Because treatments are not all observed in the same blocks, raw treatment means need not fairly compare them. Adjust treatment effects for blocks using the fitted model. The complete-block SS formulas do not apply unchanged.
:::

An incomplete block is incomplete **by design**, not a convenient description for uncontrolled missing observations. Allocate treatments and run order randomly within the design's restrictions.

## Latin Squares: Two Nuisance Factors

A $p\times p$ Latin square assigns $p$ treatments across two nuisance factors, each with $p$ levels. Every treatment appears **once in each row and once in each column**.

| | Column 1 | Column 2 | Column 3 | Column 4 |
|---|---|---|---|---|
| Row 1 | A | B | C | D |
| Row 2 | B | C | D | A |
| Row 3 | C | D | A | B |
| Row 4 | D | A | B | C |

Rows might be days and columns time windows. A letter tells us which treatment to apply at that row–column combination. There are $p^2$ runs, rather than all $p^3$ row–column–treatment combinations.

Randomly permute row assignments, column assignments, and treatment labels within an appropriate Latin-square randomization scheme. The displayed square is a layout template, not a substitute for randomization.

Let $k(i,j)$ identify the treatment in cell $(i,j)$:

$$Y_{ij}=\mu+\rho_i+\gamma_j+\tau_{k(i,j)}+\varepsilon_{ij}.$$

Under the balanced additive model, row, column, and treatment effects have sum-to-zero constraints. With $\bar Y_k^{(\mathrm{Tr})}$ the mean of the $p$ observations receiving treatment $k$,

$$\begin{aligned}
SS_R&=p\sum_i(\bar Y_{i\cdot}-\bar Y_{\cdot\cdot})^2,\\
SS_C&=p\sum_j(\bar Y_{\cdot j}-\bar Y_{\cdot\cdot})^2,\\
SS_{\mathrm{Tr}}&=p\sum_k(\bar Y_k^{(\mathrm{Tr})}-\bar Y_{\cdot\cdot})^2,\\
SS_E&=SS_T-SS_R-SS_C-SS_{\mathrm{Tr}}.
\end{aligned}$$

| Source | df | Treatment test |
|---|---|---|
| Rows | $p-1$ | |
| Columns | $p-1$ | |
| Treatment | $p-1$ | $F=[SS_{\mathrm{Tr}}/(p-1)]/MSE$ |
| Error | $(p-1)(p-2)$ | $MSE=SS_E/[(p-1)(p-2)]$ |
| Total | $p^2-1$ | |

Under the treatment null, $F\sim F_{p-1,(p-1)(p-2)}$ under independent normal common-variance errors and additivity. For the $4\times4$ square, error df is $6$. A $2\times2$ square has zero error df and cannot supply this error estimate.

The basic design cannot separately estimate general row–column, row–treatment, and column–treatment interactions. Strong interactions can invalidate the additive analysis.

## Reproduce the Complete-Block Example

```python
import numpy as np
from scipy import stats
# RCBD: rows are treatments; columns are blocks.
yb = np.array([[8, 12, 9, 14], [10, 13, 13, 16], [13, 16, 14, 18]], dtype=float)
m, b = yb.shape
grand = yb.mean()
tr_means, block_means = yb.mean(axis=1), yb.mean(axis=0)
fitted = tr_means[:, None] + block_means[None, :] - grand
ss_tr = b * np.sum((tr_means - grand)**2)
ss_block = m * np.sum((block_means - grand)**2)
ss_e = np.sum((yb - fitted)**2)
df_e = (m - 1) * (b - 1)
mse_block = ss_e / df_e
F = (ss_tr / (m - 1)) / mse_block
assert np.isclose(ss_tr + ss_block + ss_e, np.sum((yb - grand)**2))
print(f"RCBD: MSE={mse_block:.4f}, F={F:.4f}, "
      f"p={stats.f.sf(F, m-1, df_e):.8f}")
```

## Choosing the Analysis

| Design or question | Analysis | Essential check |
|---|---|---|
| Two independent quantitative groups | Pooled or Welch $t$ | Independent units; appropriate variance model. |
| Two matched conditions | Paired $t$ or within-pair randomization | Genuine pairing and independent pairs. |
| Several independent means | One-way ANOVA | Common variance and residual assumptions. |
| Specific mean combination | Contrast in the fitted model | Coefficients, SE, residual df, and testing family. |
| All pairwise mean differences | Tukey or another planned family procedure | Simultaneous rather than individual inference. |
| One nuisance factor; every treatment per block | RCBD | Randomization within blocks and additivity. |
| One nuisance factor; limited block capacity | Incomplete-block model | Connectivity, balance, and block-adjusted comparisons. |
| Two suitable nuisance factors | Latin square | One occurrence per row/column and negligible interactions. |

:::{dropdown} Self-Check Questions & Answers

**1. Why can a paired test be more precise than an independent test?**

Positive within-pair covariance reduces the variance of the difference. Pairing must reflect the actual design.

**2. Can all observations be shuffled in a blocked randomization test?**

No. Preserve the within-block assignment restrictions; otherwise you simulate an experiment that was not run.

**3. Does a significant ANOVA show every treatment differs?**

No. It only rejects equality of all means. Focused comparisons require contrasts or a multiple-comparison procedure.

**4. Is $(1,1,1)$ a contrast?**

No: the coefficients do not sum to zero. $(1,1,-2)$ is a contrast comparing the first two means with the third.

**5. Does choosing contrasts in advance remove all multiplicity concerns?**

No. A family of planned tests still needs appropriate error control if conclusions are intended to hold jointly.

**6. Why not use raw means for an incomplete-block comparison?**

Treatments may have been observed in different blocks. Their raw mean difference can include block differences.

**7. Does blocking always lower MSE?**

No. It removes block SS but also reduces residual df. It helps when the nuisance variation removed is substantial.

**8. Why does a Latin square need an additivity assumption?**

The restricted set of treatment–row–column combinations does not separately identify general interactions and pure error.
:::

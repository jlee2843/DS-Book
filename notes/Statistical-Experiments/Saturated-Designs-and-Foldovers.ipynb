# Saturated Designs & Foldovers

A very small screening experiment can examine many factors, but its estimates may mix main effects with interactions. A **foldover** adds deliberately chosen runs to change those alias relationships.

The useful question is not simply “How many more runs?” It is **which unresolved effects should the new runs distinguish?**

## What Does Saturated Mean?

A model is saturated when the number of independent fitted parameters equals the number of observations. It leaves no residual degrees of freedom.

For a main-effects model in $k$ two-level factors, there are $k+1$ coefficients including the intercept. A regular two-level screening design needs at least

$$N=2^{k-p}\geq k+1$$

runs to estimate those coefficients without main-effect aliasing. When $N=k+1$, the main-effects fit is saturated.

| Factors | Smallest power-of-two run count at least $k+1$ | Main-effects residual df |
|---|---:|---:|
| 3 | 4 | 0 |
| 4 | 8 | 3 |
| 5 | 8 | 2 |
| 7 | 8 | 0 |
| 8 | 16 | 7 |

This is a parameter-count requirement, not a promise of desirable resolution. Any residual df can contain omitted interactions when there is no genuine replication.

The maximum-factor **regular orthogonal main-effects screening designs** considered here have $N-1$ factors in $N$ runs and resolution III. “Saturated” more generally is a model-rank statement; it does not universally imply resolution III for every design or fitted model.

## Seven Factors in Eight Runs

Start with the full three-column A, B, C sign table. Assign the other four factors to its remaining product columns:

$$D=AB,\qquad E=AC,\qquad F=BC,\qquad G=ABC.$$

The intercept plus A–G occupy all eight orthogonal columns. This is a $2^{7-4}$ regular fraction with defining generators

$$I=ABD=ACE=BCF=ABCG.$$

The complete defining group contains 16 words, including I. Its shortest nonidentity words have length 3.

| Fitted main-effect column | Two-way terms in the same alias class |
|---|---|
| A | BD, CE, FG |
| B | AD, CF, EG |
| C | AE, BF, DG |
| D | AB, CG, EF |
| E | AC, BG, DF |
| F | AG, BC, DE |
| G | AF, BE, CD |

Each class also contains higher-order effects. For example,

$$E(\widehat b_D)=\beta_D+\beta_{AB}+\beta_{CG}+\beta_{EF}+\text{higher-order terms}.$$

**Do not report ordinary residual-based $t$ tests from this saturated fit.** Its residuals are zero by construction. Large contrasts may guide follow-up, but neither a ranking nor a perfect fit supplies an error estimate.

## Single-Factor Foldover

Suppose D is the priority. Add eight runs obtained by flipping **only D** in every original row, leaving the other six factor settings unchanged.

| Original fraction | Added fraction |
|---|---|
| $D=AB$ | $D=-AB$ |
| $E=AC$ | $E=AC$ |
| $F=BC$ | $F=BC$ |
| $G=ABC$ | $G=ABC$ |

A defining word containing D changes sign between the two fractions. Words without D retain their signs. In the combined design, the surviving defining words do not contain D.

For this design, D and each D-containing two-way interaction become clear of other main effects and two-way interactions. Higher-order aliases still exist; the entire experiment has not become an unaliased full factorial.

### How the New Information Appears

Ignoring higher-order terms only for this explanatory calculation,

$$E(\widehat b_A^{(1)})=\beta_A+\beta_{BD}+\beta_{CE}+\beta_{FG},$$

$$E(\widehat b_A^{(2)})=\beta_A-\beta_{BD}+\beta_{CE}+\beta_{FG}.$$

Therefore,

$$E\left[\frac{\widehat b_A^{(1)}-\widehat b_A^{(2)}}2\right]=\beta_{BD},$$

$$E\left[\frac{\widehat b_A^{(1)}+\widehat b_A^{(2)}}2\right]
=\beta_A+\beta_{CE}+\beta_{FG}.$$

The contrast of the A-column estimates isolates BD **from the other displayed low-order terms**. It does not isolate A from CE and FG.

:::{dropdown} Noiseless Example: A or BD?

Suppose the true mean is $20+2A+3BD$, with other terms zero.

In the original design, $BD=A$, so the fitted A-column coefficient is 5. After flipping D, $BD=-A$, giving an A-column coefficient of $-1$.

The average of these coefficients is 2, recovering the true A coefficient in this simplified example. Their half-difference is 3, recovering BD. With additional active aliases, the interpretation must include those terms too.
:::

## Full Sign-Reversal Foldover

A **full foldover** reverses every actual factor setting: replace each row $\mathbf x$ by $-\mathbf x$.

A word of length $j$ changes by a factor $(-1)^j$. Thus:

- Odd-length defining words change sign and no longer define the combined design.
- Even-length defining words keep their sign and remain.

The seven-factor example therefore loses its length-3 words while retaining seven length-4 words:

$$\begin{aligned}
I&=ABCG=ABEF=ACDF=ADEG\\
 &=BCDE=BDFG=CEFG.
\end{aligned}$$

The combined 16-run design has resolution IV. Main effects are clear of two-way interactions, but some two-way interactions remain aliased with one another and main effects remain aliased with higher-order terms.

![Low-order aliases before and after a D-only foldover and a full sign-reversal foldover.](../../data/Statistical-Experiments/foldover-aliases.svg)

### Reverse Settings, Not Every Generator Formula

For the seven-factor design, expressing the reversed rows in terms of their **new** base-factor values gives

$$D=-AB,\qquad E=-AC,\qquad F=-BC,\qquad G=ABC.$$

The G generator remains positive. Reversing all factors is **not** equivalent to blindly negating every dependent-factor formula: a three-factor product itself changes sign, while a two-factor product does not.

This distinction matters when writing code to construct follow-up runs. The simplest implementation is to negate every actual column of the original run table.

## What Resolution Can a Foldover Achieve?

Removing all odd defining words eliminates resolution-III aliases and yields resolution at least IV, or a full factorial if no nonidentity words remain. It need not give exactly IV in every case.

A full sign reversal cannot remove length-4 defining words, so it does not turn a resolution-IV design into resolution V. If all defining words are even, the reversed treatment rows may already be in the original fraction: extra runs then replicate that fraction rather than resolve its structural aliases.

A targeted augmentation may distinguish an important pair of interactions more efficiently than a generic foldover. Check the combined model matrix and alias sets for the specific effects of interest.

## Choosing a Follow-Up

| Objective | Useful strategy | Remaining limitation |
|---|---|---|
| Learn about D and its two-way interactions | D-only foldover in the illustrated design. | Other main-effect/two-way alias conflicts remain. |
| Separate all main effects from two-way terms | Full sign reversal of a resolution-III design. | Two-way terms may still be aliased with one another. |
| Separate a particular interaction pair | Choose runs that distinguish their columns. | Must verify the full proposed model is estimable. |
| Estimate experimental noise | Independently replicate appropriate settings. | Replication alone does not necessarily remove aliases. |

Treat the original and follow-up runs as a sequential experiment. Record any day, batch, or phase changes, randomize within practical restrictions, and include an appropriate phase effect if needed. Phase itself can be confounded with a treatment contrast in the combined layout; verify rank after adding it.

A foldover is not automatically a full replication, and additional residual df are not automatically pure error. They depend on the combined design and the terms included in the analysis.

## Verify the Foldovers in Python

The code constructs all three run tables, verifies the original main-effects model is saturated, and checks low-order alias counts directly from the columns.

```python
from itertools import combinations
import numpy as np

base = np.array([[1 if (i >> j) & 1 else -1 for j in range(3)]
                 for i in range(8)])
A, B, C = base.T
original = np.column_stack([A, B, C, A*B, A*C, B*C, A*B*C])
partial = original.copy()
partial[:, 3] *= -1  # Flip actual D settings only.
full = -original    # Reverse every actual factor setting.
pairs = list(combinations(range(7), 2))
names = list("ABCDEFG")

def pair_columns(rows):
    return np.column_stack([rows[:,i]*rows[:,j] for i,j in pairs])

for name, rows in [("Original", original),
                   ("D-only combined", np.vstack([original, partial])),
                   ("Full combined", np.vstack([original, full]))]:
    X = np.column_stack([np.ones(len(rows)), rows])
    pair_matrix = pair_columns(rows)
    count = (np.abs(rows.T @ pair_matrix) == len(rows)).sum(axis=1)
    print(name, "main-effects residual df:", len(rows)-np.linalg.matrix_rank(X))
    print("  Main/two-way alias counts:", dict(zip(names, count.tolist())))
    if name == "Full combined":
        assert np.all(count == 0)

# Verify every D-containing pair is clear of other low-order terms
# in the targeted augmentation, while allowing higher-order aliases.
rows = np.vstack([original, partial])
low_order = np.column_stack([rows, pair_columns(rows)])
for p, (i,j) in enumerate(pairs):
    if 3 in (i,j):
        col_index = 7+p
        same = np.flatnonzero(np.abs(low_order.T @ low_order[:,col_index]) == len(rows))
        assert np.array_equal(same, [col_index])

# Full reversal changes D/E/F generators, but not G=ABC.
assert np.array_equal(full[:,3], -full[:,0]*full[:,1])
assert np.array_equal(full[:,6], full[:,0]*full[:,1]*full[:,2])
print("Targeted low-order de-aliasing and full-foldover generator signs verified.")
```

## Quick Review

:::{dropdown} Questions & Answers

**Can seven factors be screened in eight regular runs?** Yes, but a saturated main-effects fit mixes main effects with interactions and has no internal residual error estimate.

**What changes in a D-only foldover?** Only the actual D setting in each row. Defining words containing D switch signs.

**What changes in a full foldover?** Every factor setting. Odd-length defining words switch signs; even-length words do not.

**Does a foldover remove all aliases?** No. Inspect the surviving defining group and the terms you intend to estimate.

**Why is G still ABC after a full sign reversal in this example?** Both G and the product of the three reversed base factors change sign.
:::

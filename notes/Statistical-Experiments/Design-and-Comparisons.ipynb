Plan a fair experiment, identify the experimental unit, and choose comparisons that answer the scientific question.

## Topics

::::::::::{grid} 1 1 2 2

:::::::::{grid-item-card} [Experimental Design](Experimental-Design.ipynb)

Identify what independently receives treatment before counting replicates. Randomization, replication, and blocking protect different parts of the comparison.
:::::::::

:::::::::{grid-item-card} [Two-Group Comparisons](Two-Group-Comparisons.ipynb)

Choose the test from the sampling design and variance assumptions. Report the size and uncertainty of the difference, not only a p-value.
:::::::::

:::::::::{grid-item-card} [Paired & Randomization Tests](Paired-and-Randomization-Tests.ipynb)

For paired data, analyze within-pair differences. Randomization tests must reproduce the assignments allowed by the actual design.
:::::::::

:::::::::{grid-item-card} [One-Way ANOVA](One-Way-ANOVA.ipynb)

The global test asks whether all treatment means agree. A rejection does not identify which pairs differ.
:::::::::

:::::::::{grid-item-card} [Contrasts & Multiple Comparisons](Contrasts-and-Multiple-Comparisons.ipynb)

State the comparison of interest first, then account for the family of comparisons being tested.

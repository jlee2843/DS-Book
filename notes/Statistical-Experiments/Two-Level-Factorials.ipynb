# Two-Level Factorial Designs

A **$2^k$ full factorial** studies $k$ factors, each at two levels, and includes all $2^k$ combinations. It is useful for screening: identify important factors and interactions before investing in a more detailed experiment.

The interaction interpretation comes from [Factorial Experiments & Interaction](./Factorial-Experiments.ipynb). Here, coding levels as $-1$ and $+1$ makes the calculations especially simple.

| Factors | Combinations at two levels | Combinations at three levels |
|---|---:|---:|
| 3 | 8 | 27 |
| 5 | 32 | 243 |
| 8 | 256 | 6,561 |

Multiply by the number of independent replicates per combination for the total run count. Two levels reduce the cost, but run counts still grow rapidly.

## Coding Natural Units

For low and high numerical settings $X_L,X_H$, code

$$x=\frac{X-(X_H+X_L)/2}{(X_H-X_L)/2}.$$

Then $X_L\mapsto-1$, $X_H\mapsto+1$, and their midpoint maps to $0$. To recover natural units,

$$X=\frac{X_H+X_L}{2}+x\frac{X_H-X_L}{2}.$$

:::{dropdown} Coding Example

For a temperature factor ranging from $40$ to $60$ degrees,

$$x=\frac{X-50}{10}.$$

Thus $40\to-1$, $50\to0$, and $60\to+1$. A coded increase of 2 is the complete low-to-high change of 20 degrees.

Categorical settings such as supplier A/B can also be labeled $-1,+1$ for comparison. Their coding does **not** imply a meaningful midpoint or numerical distance between categories.
:::

## Standard Order & Sign Columns

In standard (Yates) order, A changes sign every row, B every two rows, and C every four. A letter in a treatment label means that factor is high; $(1)$ means all factors are low.

| Label | A | B | C | AB | AC | BC | ABC |
|---|---:|---:|---:|---:|---:|---:|---:|
| $(1)$ | -1 | -1 | -1 | +1 | +1 | +1 | -1 |
| $a$ | +1 | -1 | -1 | -1 | -1 | +1 | +1 |
| $b$ | -1 | +1 | -1 | -1 | +1 | -1 | +1 |
| $ab$ | +1 | +1 | -1 | +1 | -1 | -1 | -1 |
| $c$ | -1 | -1 | +1 | +1 | -1 | -1 | +1 |
| $ac$ | +1 | -1 | +1 | -1 | +1 | -1 | -1 |
| $bc$ | -1 | +1 | +1 | -1 | -1 | +1 | -1 |
| $abc$ | +1 | +1 | +1 | +1 | +1 | +1 | +1 |

Interaction columns are products: $x_{AB}=x_Ax_B$, $x_{ABC}=x_Ax_Bx_C$. Add a column of $+1$s for the intercept.

**Standard order is for bookkeeping, not execution.** Randomize the actual run order. Running all low-C settings first could confound C with equipment drift, fatigue, or another time-related change.

## Effects as Signed Contrasts

For an effect column $s$ containing $-1,+1$, define the factorial effect as

$$\widehat E_s=\bar Y_{s+}-\bar Y_{s-}.$$

For A this compares its high and low levels, averaging over all other factors. For AB it compares the combinations with **positive product signs** against those with negative product signs. “AB positive” is not the same as “A and B both high.”

Let $r$ be the number of independent replicates at each corner and $N=r2^k$. If $T_i$ is the response total at corner $i$,

$$C_s=\sum_{i=1}^{2^k}s_iT_i,\qquad
\widehat E_s=\frac{C_s}{r2^{k-1}}.$$

Equivalently, using corner means,

$$\widehat E_s=\frac1{2^{k-1}}\sum_i s_i\bar Y_i.$$

Keep totals and means distinct: using a total-based denominator with corner means introduces an erroneous factor of $r$.

### Why Regression Coefficients Are Half the Effects

The full coded regression contains the intercept and all factor-product columns. For two factors,

$$Y=\beta_0+\beta_Ax_A+\beta_Bx_B+\beta_{AB}x_Ax_B+\varepsilon.$$

In a balanced full factorial,

$$\widehat\beta_0=\bar Y,\qquad \widehat\beta_s=\frac{\widehat E_s}{2}.$$

A coded factor changes by **two units**, from $-1$ to $+1$. This is why its averaged high-minus-low effect is twice the coefficient.

For the two-factor model, the simple A effect at B level $x_B$ is

$$\widehat\Delta_A(x_B)=2(\widehat\beta_A+\widehat\beta_{AB}x_B).$$

Consequently,

$$\text{difference of A simple effects}=4\widehat\beta_{AB}=2\widehat E_{AB}.$$

The factorial AB effect is **half the difference of differences** under this standard convention. Distinguish these quantities when switching between an interaction plot and a sign-table calculation.

:::{dropdown} Worked $2^2$ Example: Effects, Coefficients & Predictions

Use the same cell means as the preceding page, now arranged in standard order:

| Label | A | B | AB | Cell mean |
|---|---:|---:|---:|---:|
| $(1)$ | -1 | -1 | +1 | 10 |
| $a$ | +1 | -1 | -1 | 12 |
| $b$ | -1 | +1 | -1 | 14 |
| $ab$ | +1 | +1 | +1 | 22 |

$$\widehat E_A=\frac{12+22}{2}-\frac{10+14}{2}=5,$$

$$\widehat E_B=\frac{14+22}{2}-\frac{10+12}{2}=7,$$

$$\widehat E_{AB}=\frac{10+22}{2}-\frac{12+14}{2}=3.$$

Therefore,

$$\widehat Y=14.5+2.5x_A+3.5x_B+1.5x_Ax_B.$$

At $(x_A,x_B)=(+1,-1)$, the fitted mean is $14.5+2.5-3.5-1.5=12$.

The simple A effects are $2(2.5-1.5)=2$ at low B and $2(2.5+1.5)=8$ at high B. Their difference is 6, whereas the conventional AB factorial effect is 3.

Evaluating the fitted equation between corners is a **model-based interpolation**, not a direct observation. It cannot establish that the response has no curvature.
:::

## Orthogonality & ANOVA

For the balanced full factorial design matrix $X$, including all product columns,

$$X^\mathsf TX=NI.$$

Different columns have zero inner product; every column has squared length $N$. Hence,

$$\widehat{\boldsymbol\beta}=\frac1N X^\mathsf T\mathbf Y,\qquad
\operatorname{Var}(\widehat{\boldsymbol\beta})=\frac{\sigma^2}{N}I.$$

The coefficient estimates are uncorrelated under independent common-variance errors; normal errors additionally imply independence. Missing runs, unequal replication, or different weights can remove this orthogonality.

Every main effect and interaction has one df. Its sum of squares is

$$SS_s=\frac{C_s^2}{r2^k}=N\widehat\beta_s^2=\frac N4\widehat E_s^2.$$

For a fully replicated design, estimate pure error within each corner:

$$SS_E=\sum_{i=1}^{2^k}\sum_{j=1}^r(Y_{ij}-\bar Y_i)^2,\qquad
\nu_E=2^k(r-1),\qquad MSE=SS_E/\nu_E.$$

Then

$$\operatorname{SE}(\widehat\beta_s)=\sqrt{MSE/N},\qquad
\operatorname{SE}(\widehat E_s)=2\sqrt{MSE/N},$$

$$t_s=\frac{\widehat E_s}{2\sqrt{MSE/N}},\qquad
F_s=\frac{SS_s}{MSE}=t_s^2.$$

Under the corresponding zero-effect null, use $t_{\nu_E}$ or $F_{1,\nu_E}$. An individual effect interval is

$$\widehat E_s\pm t_{1-\alpha/2,\nu_E}\,2\sqrt{MSE/N}.$$

When examining many effects, distinguish individual tests from a family of claims and plan the multiplicity treatment accordingly.

:::{dropdown} Match the Coded Analysis to the Factorial ANOVA

Take three observations per cell, each with deviations $-1,0,+1$ from the cell mean. Then $N=12$, $SS_E=8$, $\nu_E=8$, and $MSE=1$.

$$SS_A=\frac{12}{4}(5^2)=75,\quad
SS_B=\frac{12}{4}(7^2)=147,\quad

These are exactly the sums of squares on the factorial ANOVA page. Each effect has SE $2/\sqrt{12}=0.5774$.

## Saturated Designs & Error Estimation

An unreplicated $2^k$ design has $2^k$ observations and $2^k$ coefficients in the full model, including the intercept. It is **saturated**:

$$\nu_E=2^k-2^k=0.$$

It fits the corner responses exactly. A perfect fit is not evidence of negligible noise or an adequate response surface.

| Option | What supplies error information? | Limitation |
|---|---|---|
| Replicate corners independently | Within-setting variation. | Requires additional runs. |
| Replicate center settings | Variation at the center. | Using it at corners assumes a common error variance. |
| Pool scientifically negligible effects into residuals | Unmodeled contrast variation. | Depends on those effects truly being negligible. |
| Use relevant external variance information | Prior or separate measurements. | Must be applicable to the present process and units. |

For example, a $2^3$ design has 8 runs. Fitting the intercept, three main effects, and three two-way interactions leaves one residual df, corresponding to ABC. This is an error estimate **only under the assumption that ABC is negligible**; it is not pure replication error.

Do not select a favorable error denominator simply to obtain significance. Preserve hierarchy: retain the lower-order terms needed by any retained interaction. A single marginal effect can be small even when that factor participates in an important interaction.

## From Screening to a Better Experiment

A screening experiment identifies directions and dependencies worth investigating. It does not automatically determine the best operating conditions.

- Use response units and interaction plots to judge practical importance, not only $p$-values or marginal boxplots.
- Choose low/high settings that are feasible and scientifically distinguishable.
- Account for participant, batch, or day when it is part of the assignment structure. Repeated trials from one participant are not automatically independent replicates.
- For quantitative factors, investigate intermediate settings and curvature when needed.
- For categorical factors, additional categories are new levels; they are not numerical midpoints.
- Validate a promising combination with additional independent runs before treating its observed advantage as established.

A reduced model or a design using fewer than all combinations introduces additional assumptions. The full-factorial sign-table formulas above should not be applied unchanged to an arbitrary subset of runs.

## Reproduce Effects & Regression

```python
import numpy as np
from scipy import stats

# Standard order: (1), a, b, ab.
A = np.array([-1, 1, -1, 1])
B = np.array([-1, -1, 1, 1])
X_corner = np.column_stack([np.ones(4), A, B, A*B])
y_corner = np.array([[9, 10, 11], [11, 12, 13],
                     [13, 14, 15], [21, 22, 23]], dtype=float)
r = y_corner.shape[1]
X = np.repeat(X_corner, r, axis=0)
y = y_corner.ravel()
N = len(y)
assert np.allclose(X.T @ X, N*np.eye(4))
coef = np.linalg.lstsq(X, y, rcond=None)[0]
residuals = y - X @ coef
df_e = N - np.linalg.matrix_rank(X)
mse = residuals @ residuals / df_e
se_effect = 2*np.sqrt(mse/N)
print(f"Intercept={coef[0]:.4f}; MSE={mse:.4f}; error df={df_e}")
for name, beta in zip(["A", "B", "AB"], coef[1:]):
    effect = 2*beta
    ss = N*beta**2
    t = effect/se_effect
    assert np.isclose(t*t, ss/mse)
    print(f"{name}: coefficient={beta:.4f}, effect={effect:.4f}, "
          f"SS={ss:.4f}, t={t:.4f}, p={2*stats.t.sf(abs(t), df_e):.8f}")
assert np.allclose(X_corner @ coef, y_corner.mean(axis=1))
print("Fitted corner means:", X_corner @ coef)
```

## Quick Review

:::{dropdown} Questions & Answers

**Why is a coded slope half the effect?** The low-to-high coded change is 2, and a balanced full factorial averages out other product terms.

**What does a positive AB sign mean?** The product of the A and B signs is positive: both low or both high.

**Does standard order control bias?** No. It organizes the design table. Randomization protects the actual run order.

**Does $R^2=1$ validate an unreplicated full model?** No. The model is saturated and has no residual df.

**Can two levels estimate a pure quadratic coefficient?** No. At every corner, $x_j^2=1$, indistinguishable from the intercept column.
:::

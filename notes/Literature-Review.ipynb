# Literature Review

Brief reviews of research on LLM evaluation, reliability and measurement.

:::{div}
:class: literature-catalog

| Paper / Review | Year | Journal |
| --- | --- | --- |
| [Quantifying Biases in LLM-as-a-Judge Evaluations](Literature/Quantifying-Biases-in-LLM-as-a-Judge-Evaluations.ipynb) **Dubois et al.** Bayesian models for five judge-evaluation questions: ordinal likelihoods, priors, posterior effects, hierarchical pooling, agreement simulations and pairwise length bias. Model choices, original figures and reported results. | 2026 | ICML (PMLR) |
| [Reliable decision support with LLMs](Literature/Reliable-Decision-Support-with-LLMs.ipynb) A framework for evaluating consistency in binary text classification applications. **Megahed et al.** 14 models, 1,350 financial news articles and five repeated classifications per model. A study of repeatability, model agreement and the gap between sentiment labeling and market prediction. | 2026 | Journal of Business Analytics |

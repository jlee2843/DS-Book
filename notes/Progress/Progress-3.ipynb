## Experiment

This report measures repeated evaluation score consistency and examines how question wording and generation temperature affect Qwen response scores and variability. The experiment used one MTRAG Govt question, task `f0d2873b877409f61da7dbdddd22d279<::>1`, turn 1, with the Original, A, B and C wordings from Progress 1.

| Experimental factor | Setting |
| --- | --- |
| Generator | Qwen3 4B |
| Prompt versions | Original, A, B, C |
| Generation temperatures | 0.3, 0.6, 0.9 |
| Conditions | 4 prompts × 3 temperatures = 12 |
| Generation repeats | 3 responses per condition |
| Generated responses | 36 |
| Evaluation repeats | 3 evaluations of each response for each metric |
| Evaluation metrics | Faithfulness and Answer Relevancy |
| Planned evaluation slots | 108 per metric; 216 across both metrics |
| Model observations | Up to 36 response averages per metric, using complete evaluation triplets |
| Evaluation summary | Each response average uses its 3 repeated evaluation scores |

Prompt and temperature were crossed: each prompt was evaluated at all three temperatures. Each generation produced a separate response from the same starting conversation state, with a distinct recorded seed. The 36 existing Qwen responses were retained for this evaluation run. For each response, its text remained unchanged across the three evaluations by a given metric. The two metrics were analyzed separately.

### Evaluation

| Metric | Implementation | Evaluation input |
| --- | --- | --- |
| Faithfulness | DeepEval with verbatim sentence splitting and sentence ID/prefix matching | Original question, generated response and fixed reference excerpts |
| Answer Relevancy | DeepEval AnswerRelevancyMetric | Original question and generated response |

All evaluations used Llama 3.1 8B locally through Ollama, with judge temperature 0 and context length 8192. The initial output limit was 2048 tokens. DeepEval version was 4.3.1. The initial run attempted three fresh evaluations per response and metric, with a new conversation for each internal model call and no reuse of completed scores between repeats. Evaluation took place on 11 November 2026.

Faithfulness used the established sentence splitter. Each sentence received an ID and a 32 character prefix. Every returned ID had to appear exactly once. Returned prefixes had to identify the corresponding source sentence uniquely and contain at least the supplied prefix length after whitespace and Unicode normalization. Longer verbatim prefixes were accepted. Verdicts were reordered by sentence ID before scoring. Reference facts were extracted anew in each evaluation. The score retained the existing DeepEval rule: both `yes` and `borderline` counted as passing. This was a sentence based adaptation of Faithfulness, not the unmodified claim extraction metric. Answer Relevancy retained its standard statement extraction and scoring procedure.

The Original question was used as the evaluator input for every prompt version. Response text was evaluated without its JSON wrapper, on the native 0 to 1 score scale. Unresolved evaluations retained a missing score and their error record; they were not converted to zero. Completed valid evaluations were retained without rerunning them. Regression and variability summaries used responses with all three valid scores.

Metric definitions: [DeepEval Faithfulness](https://deepeval.com/docs/metrics-faithfulness) and [DeepEval Answer Relevancy](https://deepeval.com/docs/metrics-answer-relevancy).

A prefix validation defect initially rejected 18 evaluations because the evaluator copied a longer verbatim prefix, such as completing a word cut off at the character limit. The corrected matching rule was applied to all 108 stored Faithfulness evaluation attempts. This recovered 18 scores without new model calls or changes to the verdicts and scoring rule. The original records were retained, and the results below use the corrected scores.

A local server interruption affected 127 evaluation slots. Their connection records were archived, and all interrupted slots were resumed with the identical model version, model digest and scoring settings. Completed model judgments, including validation failures, were retained. The reported evaluation counts use the resumed results.

The three Faithfulness evaluations for `original-0.6-2` that reached the output limit were each rerun once with a 4096 token limit. The model version, model digest, judge temperature, conversation state, prompts, sentence matching and score rule were unchanged. 0/3 retries produced valid scores; 3/3 remained failures. All earlier attempts were retained. The analysis uses one retained result per planned slot, with the retry replacing the corresponding failed attempt. There were 111 Faithfulness attempts and 108 Answer Relevancy attempts, for 219 attempts across both metrics.

## Statistical modelling

::::::{admonition} Notation toolbox
:class: note

**Indices**

| Index | Meaning | Values |
| --- | --- | --- |
| $p$ | Prompt version | Original, A, B, C |
| $t$ | Generation temperature | 0.3, 0.6, 0.9 |
| $r$ | Generated response index within a prompt × temperature condition | 1, 2, 3 |
| $j$ | Evaluation repeat index for an unchanged response | 1, 2, 3 |
| $m$ | Metric, with a separate model for each metric | Faithfulness or Answer Relevancy |
| $a$ | Prompt contrast index | A, B, C; Original is the reference |
| $b$ | Temperature contrast index | 0.6, 0.9; 0.3 is the reference |

**Scores and descriptive statistics**

| Symbol | Meaning |
| --- | --- |
| $q_{ptrjm}$ | Individual evaluation score |
| $y_{ptrm}$ | Mean of the three evaluation scores for one response; the regression outcome |
| $\bar y_{ptm}$ | Mean of the complete response averages in one condition |
| $s_{ptm}$ | Sample SD of complete response averages; includes remaining evaluation variation |
| $s_{E,ptrm}$ | Sample SD of the three evaluations of one response |

**Model terms**

| Symbol | Meaning |
| --- | --- |
| $x_{ap}$ | Indicator: 1 if prompt $p$ is variant $a$, otherwise 0 |
| $z_{bt}$ | Indicator: 1 if temperature $t$ equals category $b$, otherwise 0 |
| $x_{ap}z_{bt}$ | Product of the prompt and temperature indicators |
| $\beta_{0m}$ | Intercept: expected score for Original at temperature 0.3 |
| $\beta_{am}$ | Prompt effect: difference from Original at temperature 0.3 |
| $\gamma_{bm}$ | Temperature effect: difference from 0.3 for Original |
| $\delta_{abm}$ | Interaction effect: additional difference for prompt $a$ at temperature $b$ |
| $\epsilon_{ptrm}$ | Residual error in a response average after accounting for prompt, temperature and their interaction |


::::::

A separate linear regression was fitted for each metric when all 12 condition means were estimable. Prompt, temperature and their interaction are fixed effects. Each generated response contributes one average evaluation score to the regression.

### Response variable

For each response, the three repeated evaluation scores are averaged:

$$
y_{ptrm}=\frac{1}{3}\sum_{j=1}^{3}q_{ptrjm}.
$$

The outcome $y_{ptrm}$ remains on the original 0 to 1 score scale. With no missing scores, the analysis contains 36 response averages per metric: four prompts, three temperatures and three generated responses per condition. Individual evaluations of the same response are not treated as separate independent regression observations.

| Unit | Structure | Analysis |
| --- | --- | --- |
| Experimental condition | Prompt and temperature are crossed, giving 12 combinations | Prompt, temperature and interaction effects |
| Generated response | Three responses within each condition | One response average per regression observation |
| Repeated evaluation | Three evaluations of each unchanged response, per metric | Average score and evaluation SD for that response |

### Regression equation

Original and temperature 0.3 are the reference categories. Temperature is categorical. The indicators $x_{ap}$ and $z_{bt}$ identify the prompt and temperature contrasts.

$$
\begin{aligned}
y_{ptrm}={}&\beta_{0m}
+\sum_{a\in\{A,B,C\}}\beta_{am}x_{ap}\\
&+\sum_{b\in\{0.6,0.9\}}\gamma_{bm}z_{bt}\\
&+\sum_{a\in\{A,B,C\}}\sum_{b\in\{0.6,0.9\}}
\delta_{abm}x_{ap}z_{bt}\\
&+\epsilon_{ptrm}.
\end{aligned}
$$

The coefficients describe condition means. The interaction describes differences in the temperature effect across prompts. The residual captures differences between response averages and their condition mean, including evaluation variation remaining after averaging. It is not a measure of evaluator variation alone.

### Estimation and assumptions

The regression uses ordinary least squares (OLS). The model has one intercept, three prompt contrasts, two temperature contrasts and six interaction contrasts, giving 12 coefficients for the 12 condition means. With all 36 response averages, the residual degrees of freedom are $36-12=24$. After excluding incomplete evaluation triplets, the fitted model uses $N-12$ residual degrees of freedom when all 12 conditions remain represented.

Generated responses are treated as independent observations. Constant residual variance and approximately normal residuals are assumptions for conventional regression intervals and tests. Residual diagnostics assess these assumptions for the bounded scores. Three responses per condition limit the precision of condition comparisons.

### Measuring variability

Variability is summarized separately from the regression coefficients using sample standard deviations.

| Summary | Symbol | Interpretation |
| --- | --- | --- |
| Variation between response averages | $s_{ptm}$ | Spread of complete response averages within one prompt × temperature condition |
| Evaluation repeatability | $s_{E,ptrm}$ | Spread of the three repeated evaluation scores for one unchanged response |

The condition mean and sample SD use the $n_{ptm}$ responses with three valid evaluation scores:

$$
\bar y_{ptm}=\frac{1}{n_{ptm}}\sum_{r\in\mathcal R_{ptm}}y_{ptrm},
\qquad
s_{ptm}=\sqrt{\frac{1}{n_{ptm}-1}\sum_{r\in\mathcal R_{ptm}}(y_{ptrm}-\bar y_{ptm})^2}.
$$

The condition SD is defined when $n_{ptm}\geq 2$. With three complete responses, its divisor is $3-1=2$. The condition SD $s_{ptm}$ includes evaluation variation remaining after averaging. It does not isolate variation caused by response generation alone.

The evaluation SD for each response is:

$$
s_{E,ptrm}=\sqrt{\frac{1}{2}\sum_{j=1}^{3}(q_{ptrjm}-y_{ptrm})^2}.
$$

The evaluation SD describes repeatability under the evaluation procedure, including any stochastic preprocessing and other uncontrolled variation during evaluation. It is computed from the three repeated scores, separately from the regression residual. These individual SDs are descriptive summaries; variance components are estimated separately below. The calculations follow the standard [nested measurement summaries](https://www.itl.nist.gov/div898/handbook/mpc/section4/mpc44.htm).

### Nested variance analysis

The individual scores retain evaluations nested within responses, and responses nested within prompt × temperature conditions. Prompt and temperature are crossed. The regression above estimates condition means; the nested ANOVA below separates variation at the response and evaluation levels without adding a response term to that regression equation.

Only complete response triplets enter this analysis. With all evaluations available, $N=36$ responses and $K=12$ conditions contribute $N-K=24$ response degrees of freedom and $2N=72$ evaluation degrees of freedom.

$$
SS_{R,m}=3\sum_{p,t,r}(y_{ptrm}-\bar y_{ptm})^2,
\qquad
SS_{E,m}=\sum_{p,t,r,j}(q_{ptrjm}-y_{ptrm})^2.
$$

$$
MS_{R,m}=\frac{SS_{R,m}}{N-K},
\qquad
MS_{E,m}=\frac{SS_{E,m}}{2N}.
$$

$$
\widehat{\sigma}_{E,m}^{2}=MS_{E,m},
\qquad
\widehat{\sigma}_{R,m}^{2}=\max\left(0,\frac{MS_{R,m}-MS_{E,m}}{3}\right).
$$

The sums include complete response triplets only. The response component accounts for evaluation variation remaining in a response average. The evaluation component describes repeated scoring under the fixed evaluation procedure. The decomposition assumes independent generation runs, independent evaluation errors conditional on a response, and common variance components across conditions within a metric. The calculations follow the [nested measurement variance approach](https://www.itl.nist.gov/div898/handbook/mpc/section4/mpc44.htm) and [pooled repeatability SD](https://www.itl.nist.gov/div898/handbook/mpc/section4/mpc441.htm).

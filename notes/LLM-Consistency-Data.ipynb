# LLM Consistency Data

Data accompanying **Reliable decision support with LLMs: a framework for evaluating consistency in binary text classification applications**, by Fadel M. Megahed, Ying-Ju Chen, L. Allison Jones-Farmer, Younghwa Lee, Jiawei Brooke Wang and Inez M. Zwetsloot.

[Read the paper review](Literature/Reliable-Decision-Support-with-LLMs.ipynb) · [Author repository](https://github.com/fmegahed/llm_consistency) · [Analysis report](https://fmegahed.github.io/research/llm_consistency/llm_consistency.html)

## Dataset Structure

One input row is **one financial news article**. Models classify the same articles repeatedly so that their labels can be compared across runs and across models. These are classification experiments, not conversations or human ratings of answers.

| Dataset | Size | Meaning |
| :--- | :--- | :--- |
| Collected news | 10,000 articles | StockNewsAPI collection before experimental sampling. |
| Experiment input | 1,350 articles | 675 Negative and 675 Positive labels; verified from the published CSV. |
| Model experiment | 14 models, five runs | Repeated classifications of the same input articles. |

The `sentiment` field is the news provider’s label. It is not an independent human evaluation of each model response. The analysis also compares classifications with subsequent stock returns; these are different reference measures. [1, 2]

## Input Fields

The experimental CSV contains these ten fields. [2]

| Field | Meaning |
| :--- | :--- |
| title | News headline. |
| text | News text supplied in the dataset. |
| tickers | Stock symbol identifying the company. |
| sentiment | Provider label: Positive or Negative. |
| date | News timestamp. |
| source_name | Publisher or news source. |
| news_url | Link to the source article. |
| image_url | Link to an associated image. |
| topics | Topic metadata. |
| type | Content-type metadata. |

The model input uses the title, text and ticker. The provider sentiment label is used for comparison, rather than being shown as the answer in the prompt. [1]

## Results Files

The repository separates saved analysis outputs from the full response file. The following files are present in its results folder. [3]

| Files | Purpose |
| :--- | :--- |
| binary_analysis_df.csv; binary_ensembling_df.csv | Processed classification and ensemble analysis tables. |
| binary_pa_summary.csv; binary_reliability_metrics.csv; binary_reliability_metrics_means.csv | Agreement and reliability summaries. |
| inter_rater_reliability.csv | Agreement across models. |
| binary_sens_validity_df.csv; binary_gold_validity_df.csv; binary_validity_df.csv | Comparisons against external reference measures. |
| stock_prices_raw.csv; stock_returns_df.csv | Market data used in the return-based analysis. |

Saved outputs support inspecting the reported analysis without generating new model responses. Repeating model calls is a separate step requiring access to the relevant models. [1, 3]

## Research Focus

The paper develops a framework for checking whether LLMs can serve as consistent binary text classifiers. It distinguishes **repeatability within a model**, **agreement between models**, and **validity against an external reference**. These answer different questions: a model can give the same label repeatedly without that label being useful for the intended decision.

## Study Design and Methods

The case study asks 14 LLMs to classify financial news sentiment. Each model labels the same 1,350 articles five times, using a standardized prompt and fixed temperature settings.

| Component | Method and purpose |
| :--- | :--- |
| Sample-size planning | Adapt psychometric agreement formulas to a chosen confidence level and margin of error; use a conservative sample requirement across metrics. The planning example uses 90% confidence and a 10% margin of error. |
| Within-model reliability | Compare five repeated classifications of each article to measure how consistently a model repeats its labels. |
| Between-model reliability | Compare labels from different models to assess agreement across a panel. |
| Agreement measures | Use Conger's kappa, Fleiss' kappa, Gwet's AC1, Brennan–Prediger and Krippendorff's alpha. These account for chance agreement in different ways. |
| Invalid responses | Compare analyses that drop invalid outputs with analyses that penalize them, rather than silently treating every response as valid. |
| Validity | Compare labels with StockNewsAPI sentiment labels and, separately, actual stock excess returns. Report accuracy, sensitivity, specificity and F1. |

## Main Findings

### Repeatability and agreement

| Result | Reported value | Interpretation |
| :--- | :--- | :--- |
| Identical labels across five repeats | **88–98% of articles for most models** | Most models usually repeated their classification, but this range does not include every model. |
| DeepSeek-R1, 1.5B vs. 7B | Perfect agreement: **76.2% vs. 86%**; three agreement coefficients: approximately **0.85 vs. 0.93** | The larger DeepSeek variant was more consistent in this comparison. |
| Selected model panels | Agreement coefficients typically **above 0.70** | Adding models generally reduced agreement; panels were selected by reliability ranking, not sampled exhaustively. |

Source: Sections 4.3.1–4.3.2 and Figures 5–6. Perfect agreement is the proportion of articles receiving the same valid label in all five runs; it is not classification accuracy.

### Accuracy against StockNewsAPI sentiment labels

| Model | Accuracy | F1 |
| :--- | ---: | ---: |
| Gemma 3 1B | **0.88** | **0.868** |
| Llama 3.2 1B | **0.86** | **0.867** |
| Claude 3.5 Haiku | **0.85** | **0.85** |
| Llama 3.2 3B | **0.76** | Not stated in the narrative |
| Command R7B | **0.85** | Not stated in the narrative |
| Command R+ | **0.77** | Not stated in the narrative |

Source: Section 4.4.1 and Figure 7(a); values are reported across repeated runs. Llama 3.2 1B exceeded 3B by **10 percentage points**, and Command R7B exceeded R+ by **8 points**. These comparisons support smaller models for this task, not a universal size-performance rule.

### Prediction of actual market movements

The external criterion was whether the stock outperformed the S&P 500 **one trading day after publication**.

| Measure | Reported result |
| :--- | :--- |
| Precision across models | **0.50–0.54** |
| F1 across models | **0.49–0.54** |
| Claude 3.7 Sonnet | Sensitivity **0.55**, specificity **0.48** |
| Command R+ | Sensitivity **0.45**, specificity **0.58** |

Source: Section 4.4.2 and Figure 7(b). Results were near chance despite strong repeatability and agreement with the API's sentiment labels. The study therefore separates consistent labeling from useful prediction.

## Limitations

| Limitation | Consequence for interpretation |
| :--- | :--- |
| One domain and two labels | Financial sentiment results do not establish performance on other domains, multiclass labels or graded ratings. |
| Fixed prompt and settings | The study does not systematically measure sensitivity to alternative prompts or temperatures. |
| Imperfect validation references | StockNewsAPI uses heuristic sentiment labels, while short-horizon returns contain substantial noise. Neither is a definitive ground truth for sentiment quality. |
| Output parsing choices | Extracting labels from misformatted responses can affect results; stricter formatting or different invalid-response policies may give different estimates. |
| Text-only evaluation | Multimodal tasks were not tested. |

## Relevance to LLM Evaluation

This paper provides a useful design for repeated scoring and agreement analysis. Its binary classification results should not be treated as direct validation of multi-turn answer judging or ordinal quality scores. Those applications require their own rating design and validity checks.

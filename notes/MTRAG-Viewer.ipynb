:::{dropdown} Display fields

Message, response and passage text is preserved from commit `53fef6f5917c7e10c436f697e69a955dbc795d81`. The display files omit author IDs, timestamps, retrieval query internals and automatic metric annotations. Turn numbers are converted to integers.

Human ratings belong to the 159-task reference subset. Their order within a metric does not identify the same rater across metrics. For the original schema and inspection code, see the [dataset guide](MTRAG-Inspection.ipynb).

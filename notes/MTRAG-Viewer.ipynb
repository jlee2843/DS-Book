## Browse the Data

Choose a collection or search question text, history and task IDs. Select a conversation and turn, then switch evidence settings to compare the passages supplied to the model. **Human ratings** shows the released candidate responses and individual scores for the 159-task reference subset.

## About This View

:::{admonition} Definition · Supplied history
:class: note

Each task includes fixed earlier user and agent messages. The current reference answer is separate. Moving to the next turn displays the released history, rather than a conversation generated by a new model during this visit.
:::

:::{dropdown} Source, display fields and reproducibility

Data comes from [IBM's MTRAG human release](https://github.com/IBM/mt-rag-benchmark/tree/53fef6f5917c7e10c436f697e69a955dbc795d81/mtrag-human), pinned to commit `53fef6f5917c7e10c436f697e69a955dbc795d81`. The viewer preserves the source message, response and passage text. It omits author identifiers, timestamps, retrieval query internals and non-human metric annotations, and converts turn numbers to integers for browsing. Display files are projections, not the original complete records.

Reference + RAG contains 436 tasks; a task outside that subset displays an explicit unavailable state. Human ratings are available for 159 tasks under the **reference** setting. Missing annotations remain missing, and individual ratings are shown without treating them as independent answers. Their order within a metric does not identify the same rater across metrics.

Original data is published by IBM under the repository's [Apache 2.0 license](../data/Research/mtrag/LICENSE.txt). Credit: the MTRAG authors; [benchmark paper](https://aclanthology.org/2025.tacl-1.36/). The [dataset guide](MTRAG-Inspection.ipynb) contains the field dictionary, inspection results, caveats and reproducible Python workflow.

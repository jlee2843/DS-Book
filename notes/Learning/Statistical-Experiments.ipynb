# Statistical Experiments

Experimental design connects the scientific question to treatment assignment, independent replication, and the analysis of observed responses. The design determines which comparisons the data can support.

Begin with Design & Mean Comparisons. Use blocking to account for nuisance variation, factorial designs to study interacting factors, response surfaces to explore optimization, and random-effects models for grouped or nested observations.

Correlated predictors and flexible models can produce unstable coefficient estimates. Regularization constrains a fit to trade some bias for greater stability.

Start with the overview and least-squares geometry, then examine collinearity through the spectrum of the design matrix. The ridge notes introduce the penalty and explain how to choose its strength.

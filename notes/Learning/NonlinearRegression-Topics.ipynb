Flexible regression represents a relationship with basis functions and controls its roughness. The aim is to capture structure without fitting every fluctuation in the sample.

Start with the overview and function spaces, then build spline bases. Penalized splines connect smoothness to estimation; the final topic covers tuning and uncertainty.

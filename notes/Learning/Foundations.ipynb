# Foundations

These notes cover the ideas used throughout statistical learning: how to measure prediction quality, recognize overfitting, and prepare data before fitting a model.

Start with Basic for evaluation metrics and model behaviour. Use Preprocessing when deciding how to encode, scale, or transform inputs. Fit data-dependent transformations on the training set before applying them to new observations.

# Nonlinear Regression

Chapter 3 asks how to estimate a curve when a straight line is too restrictive. Simply allowing every possible function does not solve the problem: many curves pass through the same observations and behave very differently between them. We need to specify both the functions we will consider and how strongly we discourage unnecessary variation.

A basis expansion makes the curve flexible. A roughness penalty controls how much of that flexibility the data can use. This is the same balance between fit and stability that motivated ridge regression, now applied to functions.

:::{admonition} A curve can be nonlinear in the input and linear in its coefficients
:class: important

For $f(x)=\beta_0+\beta_1x+\beta_2x^2$, the input follows a curved relationship, but estimating the coefficients is still a linear regression problem. The columns of the design matrix are now $1$, $x$, and $x^2$.
:::

## The model throughout this chapter

We observe $Y_i=f(x_i)+\varepsilon_i$ for $i=1,\ldots,n$, with $E(\varepsilon_i\mid X)=0$. For covariance calculations, assume independent errors with common variance $\sigma^2$. Normal errors are needed for the exact Gaussian sampling statements, rather than for the least squares minimization itself.

Write a candidate curve as

$$f(x)=\sum_{j=1}^{m}b_j(x)\beta_j=b(x)^\top\beta.$$

Here $m$ counts basis functions, while $n$ counts observations. The design matrix has entries $B_{ij}=b_j(x_i)$. Once the basis is fixed, the fitted curve is obtained by estimating $\beta$.

This chapter uses the objective $\lVert y−B\beta\rVert^2+\lambda\beta^\top S\beta$. Any scale factor involving the noise variance is absorbed into $\lambda$. Keep this convention in mind when comparing the handwritten lecture equations.

The earlier [ridge chapter](Regularization-and-Penalized-Regression.ipynb) supplies the matrix algebra. The new ingredient is $S$, which measures the roughness of the curve rather than the size of every coefficient equally.

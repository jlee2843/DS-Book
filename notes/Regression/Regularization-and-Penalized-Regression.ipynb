# Regularization & Penalized Regression

Least squares fits a linear combination of inputs by minimizing residual error. When predictors are highly correlated or numerous, the coefficients can become unstable. Regularization adds a constraint or penalty to stabilize the fit, accepting some bias in exchange for reduced variance.

:::{admonition} Why regularize?
:class: important

Accept a little bias to reduce sensitivity to noise. The aim is lower prediction error, not the smallest possible training residuals.
:::

## From Fit to Penalized Fit

Let $X\in\mathbb R^{n\times p}$ be a design matrix and $y\in\mathbb R^n$ the response vector. A column can be an original predictor, an indicator, or a transformation such as $x^2$. **Linear regression is linear in its coefficients**, not necessarily in the original inputs.

| Method | Objective | Main effect |
|---|---|---|
| Ordinary least squares | $\min_\beta\ \lVert y-X\beta\rVert^2$ | Finds the closest fitted vector in the column space of $X$. |
| Ridge, all coefficients penalized | $\min_\beta\ \lVert y-X\beta\rVert^2+\lambda\lVert\beta\rVert^2$ | Discourages large coefficients and stabilizes weakly identified directions. |

Here $\lambda\geq0$ controls regularization. Larger $\lambda$ places more weight on the penalty. It is a tuning parameter, not a coefficient estimated by the unpenalized normal equations.

### Notation & Conventions

- $p$ counts columns of the displayed design matrix, including an intercept if present.
- $r=\operatorname{rank}(X)$ counts linearly independent columns.
- $s_j$ denotes a singular value of $X$; $s_j^2$ is an eigenvalue of $X^\top X$.
- Formulas with $X^\top X+\lambda I$ penalize every displayed coefficient. The practical unpenalized-intercept convention is explained separately.
- The objective uses **sum of squared errors**. If software instead uses mean squared error, the numerical penalty parameter needs rescaling.

## What Regularization Changes

Ridge suppresses directions in which small changes in the responses could cause large changes in the least-squares coefficients. It can improve prediction error even when least squares is unbiased.

It does not guarantee that every individual coefficient decreases monotonically in magnitude, that every value of $\lambda$ improves prediction, or that highly correlated predictors become separately interpretable. Predictions and uncertainty still depend on the design and the target inputs.

:::{dropdown} Self-Check

**Why not minimize training error alone?**

Added flexibility can fit noise as well as signal.

**Does an $x^2$ column make least squares nonlinear in $\beta$?**

No. It remains a linear combination of fixed columns.

**What changes when we increase $\lambda$?**

The fit accepts a larger residual error in return for a smaller coefficient norm; effective degrees of freedom decrease.

**Must an intercept be shrunk toward zero?**

No. It is usually left unpenalized, with predictors centered before fitting the slopes.
:::

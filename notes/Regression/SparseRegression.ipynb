# Sparse Regression and Trees

Sparse regression addresses problems with many candidate predictors. Sometimes there are more columns than observations; sometimes there are simply too many plausible terms to inspect individually. The main approaches here reduce complexity in different ways. LASSO keeps a sparse collection of coefficients. A regression tree divides the input space into a small collection of regions.

## A consistent objective

Throughout these LASSO pages, $p$ counts slopes and the intercept is unpenalized. After centering the training response and predictors, use

$$Q_\lambda(\beta)=\frac12\lVert y−X\beta\rVert^2+\lambda\sum_{j=1}^{p}|\beta_j|.$$

The factor $1/2$ removes the factor 2 from the derivative of the residual loss. If the objective instead uses $\lVert y−X\beta\rVert^2+\lambda_{\mathrm{sum}}\lVert\beta\rVert_1$, the same fit has $\lambda_{\mathrm{sum}}=2\lambda$. If it uses $\lVert y−X\beta\rVert^2/(2n)+\lambda_{\mathrm{mean}}\lVert\beta\rVert_1$, the same fit has $\lambda=n\lambda_{\mathrm{mean}}$, for otherwise identical data and preprocessing.

The lecture and companion notes switch scales in places. All derivatives and thresholds below follow the displayed half sum convention.

## Prediction and variable discovery

A variable being selected means that it contributes to this fitted procedure at this penalty, with these other predictors and this sample. It does not establish a causal effect or prove that the other variables are irrelevant. Likewise, a short tree can be readable without being stable across samples.

The validation principles from [Predictive Modelling](Predictive-Modelling.ipynb) remain essential. Choosing a sparse or small model is a way to control complexity, not a substitute for measuring prediction error.

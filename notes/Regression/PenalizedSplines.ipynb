# Penalized Splines

A curve can fit every observed value and still oscillate wildly between observations. Penalized splines discourage that behavior by charging for curvature, rather than merely counting basis coefficients.

## Measuring roughness

For a sufficiently regular function, define

$$J(f)=\int_a^b\{f''(u)\}^2\,du.$$

Squaring prevents positive and negative curvature from cancelling. Integrating accumulates bending over the interval. A straight line has $f''=0$ and pays no penalty, regardless of its intercept or slope.

A smoothing spline minimizes

$$\sum_{i=1}^{n}\{y_i−f(x_i)\}^2+\lambda J(f).$$

For distinct ordered inputs and a positive smoothing parameter, the standard unrestricted boundary formulation has a natural cubic spline solution with knots at the observed inputs. A penalized regression spline instead begins with a chosen finite basis, often using fewer knots. Both use the same fit versus roughness idea, but their available function spaces need not coincide.

## Why the natural cubic spline is enough

Fix the fitted values of any competing function $f$ at the observations. Let $g$ be the natural cubic spline interpolating those same values and write $h=f−g$. Then $h(x_i)=0$.

Expanding the penalty gives

$$J(f)=J(g)+2\int_a^b g''(u)h''(u)\,du+J(h).$$

Integration by parts on each knot interval makes the cross term zero. Natural boundary conditions remove the outer boundary terms; continuity and $h(x_i)=0$ remove the remaining knot terms. Consequently $J(f)\geq J(g)$.

Replacing $f$ by $g$ leaves the residual sum of squares unchanged and cannot increase roughness. This is why searching over natural cubic splines loses nothing in the smoothing problem.

## Turning curvature into a matrix

Write $f(x)=b(x)^\top\beta$. Since $f''(x)=b''(x)^\top\beta$,

$$J(f)=\beta^\top S\beta,\qquad S_{jk}=\int_a^b b_j''(u)b_k''(u)\,du.$$

The matrix $S$ is symmetric and positive semidefinite. Its entries can be computed interval by interval because the derivatives are polynomials within each knot interval.

The coefficient objective becomes

$$L(\beta)=\lVert y−B\beta\rVert^2+\lambda\beta^\top S\beta.$$

Differentiation gives

$$−2B^\top(y−B\beta)+2\lambda S\beta=0,$$

and hence

$$\widehat\beta_\lambda=(B^\top B+\lambda S)^{−1}B^\top y.$$

This inverse exists when no nonzero coefficient direction is invisible both to the data and to the penalty. Equivalently, $\operatorname{null}(B)\cap\operatorname{null}(S)=\{0\}$. In computation, solve the system rather than explicitly forming an inverse.

## What a large penalty leaves behind

For a full spline basis containing all linear functions, the nullspace of $S$ has dimension two. It represents the intercept and slope. Larger $\lambda$ suppresses departures from a straight line, not the line itself.

If the data identify both unpenalized directions, the limit as $\lambda\to\infty$ is the least squares straight line. It is not generally the zero function. At small $\lambda$, the fit approaches the unpenalized fit available in the chosen spline basis; that fit interpolates only if the basis and design allow it.

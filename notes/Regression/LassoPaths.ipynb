# Solution Paths and Selection

A solution path records how the fitted coefficients change as the penalty varies. It makes fitting many candidate models efficient, but it does not tell us which candidate predicts best. Validation supplies that separate decision.

## Starting from the zero slope model

With centered data and $\beta=0$, the residual is $y$. The zero vector satisfies the LASSO optimality conditions exactly when $|x_j^\top y|\leq\lambda$ for every $j$. Therefore,

$$\lambda_{\max}=\max_j|x_j^\top y|.$$

Every penalty at least this large has zero as an optimal slope vector. An unpenalized intercept remains at the training mean. Under a mean loss convention, the threshold is divided by $n$.

## Warm starts

Choose a decreasing sequence beginning at $\lambda_{\max}$. Solve the first problem, then use its coefficients as the starting point for the next penalty. Neighboring problems often have similar solutions, so this avoids repeatedly starting far from the answer.

Warm starts are computational devices. Each candidate must still converge to the required tolerance. They do not replace validation and do not establish that coefficients enter in a fixed ranking of their original correlations with the response.

## Why the path is piecewise linear

Suppose the active set $A$ and its sign vector $s_A$ remain fixed over a penalty interval, and $X_A$ has full column rank. The active optimality equations are

$$X_A^\top(y−X_A\beta_A)=\lambda s_A.$$

Rearranging,

$$\beta_A(\lambda)=(X_A^\top X_A)^{−1}X_A^\top y−\lambda(X_A^\top X_A)^{−1}s_A.$$

Within this interval, the coefficients are affine functions of $\lambda$. A segment ends when an active coefficient reaches zero or an inactive variable reaches its residual correlation threshold. The set and signs can then change.

A coefficient may leave and later reenter. The number of selected variables is not guaranteed to increase at every step as the penalty decreases. Degenerate designs also require care in choosing among coefficient solutions; the displayed inverse assumes an independent active design.

## Reading a coefficient plot

A flat zero segment identifies penalties where a coefficient is absent. A change in slope marks a change in the active configuration. Correlated columns can alter one another's coefficient paths without any statistical interaction term being present.

Piecewise linearity is with respect to $\lambda$ itself. If the horizontal axis is $\log\lambda$, a straight segment in $\lambda$ will generally appear curved. Axis labels matter when comparing the derivation with a software plot.
